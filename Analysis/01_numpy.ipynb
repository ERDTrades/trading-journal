```python

import numpy as np
from journal import Journal

import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import os

os.chdir(PROJECT_ROOT)

journal = Journal()
```

```python
class numpy_statistics():
    def __init__(self, trades):
        self.trades = trades

    def pnl_in_depth(self):
        pnl = []
        for trade in self.trades:
            pnl.append(trade.pnl)
        return pnl
```

```python
numpy_stats = numpy_statistics(journal.trades)

first_array = numpy_stats.pnl_in_depth()

pnl_array = np.array(first_array)


display(pnl_array.mean())
display(pnl_array.std())
display(pnl_array.sum())
display(pnl_array.max())
display(pnl_array.min())

# jupyther is just a lab
# If something works Just create a simmilar module and copy that there
# You want to have seperate module with working classes etc.
# this is just a "workstation"
# So copy that working function and class create a new numpy module
# and from there you can import that to menu for 9th option
```
